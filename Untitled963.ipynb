```python
import xarray as xr
import numpy as np

# ==========================================
# 1. Load NetCDF file
# ==========================================
file_path = "EMEP_yearly_2015.nc"   # change if needed
ds = xr.open_dataset(file_path)

print("\n================ DATASET SUMMARY ================\n")
print(ds)

# ==========================================
# 2. Coordinates information
# ==========================================
print("\n================ COORDINATES ================\n")
for coord in ds.coords:
    print(f"\n--- Coordinate: {coord} ---")
    print(ds[coord])
    print("Attributes:", ds[coord].attrs)

# ==========================================
# 3. Data variables
# ==========================================
print("\n================ DATA VARIABLES ================\n")
for var in ds.data_vars:
    print(f"\n--- Variable: {var} ---")
    print(ds[var])
    print("Attributes:", ds[var].attrs)

# ==========================================
# 4. Global attributes
# ==========================================
print("\n================ GLOBAL ATTRIBUTES ================\n")
print(ds.attrs)

# ==========================================
# 5. Check for bounds variables
# ==========================================
print("\n================ CHECK FOR BOUNDS VARIABLES ================\n")
print("All variables in file:")
print(list(ds.variables))

possible_bounds = [v for v in ds.variables if "bnd" in v.lower() or "bound" in v.lower()]
print("\nPossible bounds variables:", possible_bounds)

# ==========================================
# 6. Grid spacing test
# ==========================================
print("\n================ GRID SPACING TEST ================\n")

if "lat" in ds.coords:
    lat = ds["lat"].values
    dlat = np.diff(lat)
    print("Latitude range:", lat.min(), "to", lat.max())
    print("Mean latitude spacing:", np.mean(dlat))
    print("First 5 latitude differences:", dlat[:5])

if "lon" in ds.coords:
    lon = ds["lon"].values
    dlon = np.diff(lon)
    print("\nLongitude range:", lon.min(), "to", lon.max())
    print("Mean longitude spacing:", np.mean(dlon))
    print("First 5 longitude differences:", dlon[:5])

# ==========================================
# 7. If bounds exist → verify centre logic
# ==========================================
print("\n================ CENTRE vs BOUNDS TEST ================\n")

if "lat_bnds" in ds.variables:
    lat_bnds = ds["lat_bnds"].values
    lat_centres = ds["lat"].values
    print("First latitude centre:", lat_centres[0])
    print("First latitude bounds:", lat_bnds[0])
    print("Midpoint of bounds:", lat_bnds[0].mean())

if "lon_bnds" in ds.variables:
    lon_bnds = ds["lon_bnds"].values
    lon_centres = ds["lon"].values
    print("\nFirst longitude centre:", lon_centres[0])
    print("First longitude bounds:", lon_bnds[0])
    print("Midpoint of bounds:", lon_bnds[0].mean())

print("\n================ DONE ================\n")

print(ds["lon"])
```

```python

import xarray as xr
import pandas as pd
import numpy as np
from sklearn.neighbors import BallTree

# --- Load Data ---
nc_file_path = "EMEP_yearly_2015.nc"
csv_file_path = "yearly_O3_2015.csv"

# --- Load NetCDF Data ---
nc_data = xr.open_dataset(nc_file_path)

lon_values = nc_data["lon"].values
lat_values = nc_data["lat"].values
o3_grid = nc_data["SURF_ug_O3"].values.squeeze()

# --- Create grid points (lat, lon) ---
lon_grid, lat_grid = np.meshgrid(lon_values, lat_values)

grid_points_deg = np.column_stack((
    lat_grid.ravel(),
    lon_grid.ravel()
))
grid_points_rad = np.radians(grid_points_deg)

o3_values = o3_grid.ravel()

# --- Load Station Data ---
station_data = pd.read_csv(csv_file_path)

station_lons = station_data["Longitude"].values
station_lats = station_data["Latitude"].values

station_coords_deg = np.column_stack((
    station_lats,
    station_lons
))
station_coords_rad = np.radians(station_coords_deg)

# --- Nearest Grid Cell Matching (Haversine) ---
tree = BallTree(grid_points_rad, metric="haversine")
distances_rad, indices = tree.query(station_coords_rad, k=1)

nearest_indices = indices.flatten()
nearest_grid_coords = grid_points_deg[nearest_indices]   # (lat, lon)
nearest_o3_values = o3_values[nearest_indices]

# --- Create Output DataFrame (KEEP ALL ORIGINAL FEATURES) ---
df_matched = station_data.copy()

df_matched["nearest_grid_lon"] = nearest_grid_coords[:, 1]
df_matched["nearest_grid_lat"] = nearest_grid_coords[:, 0]
df_matched["grid_cell_index"] = list(
    zip(nearest_grid_coords[:, 1], nearest_grid_coords[:, 0])
)
df_matched["nearest_SURF_ug_O3"] = nearest_o3_values

# --- Save to CSV ---
df_matched.to_csv("baseO3nearest_grid.csv", index=False)

print("Done. File saved as 'baseO3nearest_grid.csv'")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score

# Load data
df = pd.read_csv("baseO3nearest_grid.csv")

observed = df["Average"]
modeled = df["nearest_SURF_ug_O3"]

# Compute global R²
r2 = r2_score(observed, modeled)

# Unique station types
types = df["Type"].unique()

plt.figure(figsize=(7,7))

# Plot each type separately
for t in types:
    subset = df[df["Type"] == t]
    plt.scatter(
        subset["nearest_SURF_ug_O3"],
        subset["Average"],
        label=t,
        alpha=0.7
    )

plt.xlabel("EMEP O₃ (μg/m³)")
plt.ylabel("Observed O₃ (μg/m³)")
plt.title("Observed vs EMEP O₃ (by Station Type)")

# Add R² inside plot
plt.text(
    0.05,
    0.95,
    f"$R^2$ = {r2:.3f}",
    transform=plt.gca().transAxes,
    fontsize=12,
    verticalalignment='top'
)

plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
from scipy.stats import linregress

# Load data
df = pd.read_csv("baseO3nearest_grid.csv")

observed = df["Average"]
modeled = df["nearest_SURF_ug_O3"]

# Global regression (modeled → observed)
slope, intercept, r_value, p_value, std_err = linregress(modeled, observed)
r2 = r_value**2

# Create figure
plt.figure(figsize=(7,7))

# Scatter by station Type
types = df["Type"].unique()

for t in types:
    subset = df[df["Type"] == t]
    plt.scatter(
        subset["nearest_SURF_ug_O3"],
        subset["Average"],
        label=t,
        alpha=0.7
    )

# Regression line
x_vals = np.linspace(modeled.min(), modeled.max(), 100)
y_vals = intercept + slope * x_vals
plt.plot(x_vals, y_vals)

# Labels
plt.xlabel("EMEP O₃ (μg/m³)")
plt.ylabel("Observed O₃ (μg/m³)")
plt.title("Observed vs EMEP O₃ (by Station Type)")

# Add equation and R² inside plot
plt.text(
    0.05,
    0.95,
    f"$y = {slope:.2f}x + {intercept:.2f}$\n$R^2 = {r2:.3f}$",
    transform=plt.gca().transAxes,
    fontsize=11,
    verticalalignment='top'
)

plt.legend()
plt.tight_layout()
plt.show()
```
